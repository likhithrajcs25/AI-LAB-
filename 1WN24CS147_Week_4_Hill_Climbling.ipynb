```python

def cost(state):
    conflicts = 0
    n = len(state)

    for i in range(n):
        for j in range(i + 1, n):
            if abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1

    return conflicts
def hill_climbing():
    current = [3, 1, 2, 0]

    while True:
        current_cost = cost(current)
        print("State:", current, "Cost:", current_cost)

        if current_cost == 0:
            print("Solution found!")
            return

        best = current[:]
        best_cost = current_cost

        for i in range(4):
            for j in range(i + 1, 4):
                neighbour = current[:]
                neighbour[i], neighbour[j] = (
                    neighbour[j], neighbour[i]
                )

                neighbour_cost = cost(neighbour)

                if neighbour_cost < best_cost:
                    best = neighbour
                    best_cost = neighbour_cost

        if best_cost >= current_cost:
            print("No better neighbour found.")
            return

        current = best

hill_climbing()
```
